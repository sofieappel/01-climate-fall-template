* Hotter temperatures in Colorado are leading to more frequent and severe droughts in the state. Typicall, there is a recovery period from a drought with increased precipitation that helps to replenish the lack of water during the drought, but projections suggest that these recovery times will be less beneficial as the state continues to warm (Bolinger, 2024).
* Increased frequency and severity in droughts is leading to larger and more frequent wildfires. The USDA Forest Service projects, "the annual area burned in the western United States could increase two to six times" (Karau, 2024).
* As one of the main reasons for tourists in Colorado, snowpack is decreasing with climate change, and the season for snow is shortening. The 21st century has seen between 3% and 23% less snowpack than it did from 1951-2000 (Bolinger, 2024).
